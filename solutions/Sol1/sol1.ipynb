```python
year = int(input("Enter a year: "))
if year%400==0:
  print("Leap year")
elif (year%4==0) and (year%100!=0):
  print("Leap year")
else:
  print("Not a leap year")
```

```python
value = float(input("Enter temperarure: "))
unit = input("Enter unit (F or C): ")
if unit=='C' or unit=='c':
  value=(value*9/5) + 32
  print("Temperature in Celsius: ", value)
elif unit=='F'or unit=='f':
  value=(value-32)*(5/9)
  print("Temperature in Fahrenheit: ", value)
```
