```python
import requests
from bs4 import BeautifulSoup
import csv

# Step 1: Fetch the HTML page
URL = "https://realpython.github.io/fake-jobs/"
page = requests.get(URL)

# Step 2: Parse with BeautifulSoup
soup = BeautifulSoup(page.content, "html.parser")

# Step 3: Find the main container
results = soup.find(id="ResultsContainer")

# Step 4: Find all job cards
job_cards = results.find_all("div", class_="card-content")

# Step 5: Extract data
jobs = []
for card in job_cards:
    title = card.find("h2", class_="title").text.strip()
    company = card.find("h3", class_="company").text.strip()
    location = card.find("p", class_="location").text.strip()
    date = card.find("time").text.strip()
    link = card.find_all("a")[1]["href"]

    jobs.append([title, company, location, date, link])

# Step 6: Save to CSV
csv_filename = "jobs.csv"
with open(csv_filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Title", "Company", "Location", "Date", "Apply_URL"])
    writer.writerows(jobs)

```

```python
# Install dependencies
!pip install requests beautifulsoup4 pandas

import requests
from bs4 import BeautifulSoup
import csv

# Step 1: Fetch trending page
url = "https://github.com/trending/python?since=daily"  # daily trending
headers = {"User-Agent": "Mozilla/5.0"}  # pretend to be browser
page = requests.get(url, headers=headers)

# Step 2: Parse with BeautifulSoup
soup = BeautifulSoup(page.content, "html.parser")

# Step 3: Find all repo cards
repos = soup.find_all("article", class_="Box-row")
print("Total repos found:", len(repos))

# Step 4: Extract details
data = []
for repo in repos:
    # Repository name + URL
    repo_tag = repo.h2.a
    name = repo_tag.text.strip().replace("\n", "").replace(" ", "")
    repo_url = "https://github.com" + repo_tag["href"]

    # Description (may be missing)
    desc_tag = repo.find("p", class_="col-9")
    description = desc_tag.text.strip() if desc_tag else ""

    # Language (may be missing)
    lang_tag = repo.find("span", itemprop="programmingLanguage")
    language = lang_tag.text.strip() if lang_tag else ""

    # Stars today
    stars_today_tag = repo.find("span", class_="d-inline-block float-sm-right")
    stars_today = stars_today_tag.text.strip().replace("stars today", "").strip() if stars_today_tag else "0"

    # Total stars
    stars_tag = repo.find("a", href=lambda x: x and x.endswith("/stargazers"))
    stars = stars_tag.text.strip() if stars_tag else "0"

    # Forks
    forks_tag = repo.find("a", href=lambda x: x and x.endswith("/network/members"))
    forks = forks_tag.text.strip() if forks_tag else "0"

    # Contributors (from <img alt="@username">)
    contributors = [img['alt'].replace("@", "") for img in repo.find_all("img", class_="avatar")]

    # Append data row
    data.append([name, description, language, stars_today, stars, forks, ", ".join(contributors), repo_url])


# Step 5: Save to CSV
filename = "trending_repos.csv"
with open(filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Repository", "Description", "Language", "Stars Today", "Total Stars", "Forks", "Contributors", "URL"])
    writer.writerows(data)



```
