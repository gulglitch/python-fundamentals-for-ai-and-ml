```python
def count_vowels_and_unique_consonants(myStr):
    vowelCount = 0
    unique_consonants = []  # Use a list

    for char in myStr:
        if char.lower() in 'aeiou':
            vowelCount += 1
        elif char.isalpha():  # Check if the character is a letter
            unique_consonants.append(char.lower())  # Add lowercase consonant to the list

    consonantsCount = len(unique_consonants)
    return vowelCount, consonantsCount

myStr = input("Enter a string: ")
vowels, consonants = count_vowels_and_unique_consonants(myStr)
print( "( " + str(vowels) + " , " + str(consonants) + " )")
```

```python
myList_str = input("Enter a list of elements separated by spaces: ")
myList = myList_str.split() # Split the input string into a list
k = int(input("Enter k: "))

# each element shifts to the right and last k elements wrap around
def rotate_list(myList, k):
  # Calculate the effective rotation amount
  k = k % len(myList)
  rotated_list = myList[-k:] + myList[:-k]
  return rotated_list

rotated_result = rotate_list(myList, k)
print(" ".join(rotated_result)) # Join the rotated list elements back into a string for printing
```

```python
nums = input("Enter a sorted list of numbers separated by spaces: ")
target = int(input("Enter the target number: "))
nums_list = nums.split()

def binarySearch(nums_list, target):
    left = 0
    right = len(nums_list) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums_list[mid] == target:
            return mid
        elif nums_list[mid] < target:
            left = mid + 1
        else:
            right = mid - 1 # Corrected the logic here

    return -1

result = binarySearch(nums_list, target)

if result != -1:
    print("Found at index:", result)
else:
    print("Not found")
```

```python
data = [(1,2), (3,1), (5,0)]

def sort_tuple(data):
  #find no of elements in tuple
  n=len(data)
  for i in range(n):
    for j in range(0,n-i-1):
      # Compare based on the second element of the tuple
          if data[j][1] > data[j + 1][1]:
             # Swap the tuples
              data[j], data[j+1] = data[j + 1], data[j]
    return data

sorted_data = sort_tuple(data)
print(sorted_data)
```

```python
scores = {
'Alice': 85,
'Bob': 58,
'Charlie': 60,
'David': 45,
'Eva': 70
}

def filter_scores(scores):

  filtered_dict = {}

  for name, score in scores.items():
    if score >= 60:
      filtered_dict[name] = score
  return filtered_dict

filtered_scores = filter_scores(scores)
print(filtered_scores)
```

```python
#data is a list of tuples
data= [(1,'a'), (2,'b'), (1,'c')]

def group_by_first_element(data):

    #empty dictionary to store result
    grouped_data = {}
    for first, second in data:
        if first not in grouped_data:
            #create a new entry with an empty list
            grouped_data[first] = []
        #append the second value to the list (whether it is empty or already contains a value)
        grouped_data[first].append(second)
    return grouped_data

grouped_result = group_by_first_element(data)
print(str(grouped_result))
```

```python
dict1 = {'apple': 10, 'banana': 5, 'orange': 7}
dict2 = {'banana': 3, 'orange': 2, 'grape': 4}

# Expected Output:
# {'apple': 10, 'banana': 8, 'orange': 9, 'grape': 4}

def merge_dictionaries(dict1, dict2):

  merged_dict = dict1.copy()

  for key, value in dict2.items():
    if key in merged_dict:
      merged_dict[key] += value
    else:
      merged_dict[key] = value
  return merged_dict

merged = merge_dictionaries(dict1, dict2)
print(merged)
```

```python
def is_palindrome(word):
  #Checks if a word is a palindrome.
  return word == word[::-1]

def find_palindrome_pairs(words):

  palindrome_strings = []
  n = len(words)
  for i in range(n):
    for j in range(n):
      if i != j:  # Ensure distinct words
        concat_word = words[i] + words[j]
        if is_palindrome(concat_word):
          palindrome_strings.append(concat_word)
  return palindrome_strings

# Dynamic input
words_str = input("Enter a list of unique strings separated by spaces: ")
words = words_str.split()

result = find_palindrome_pairs(words)
print(result)
```
