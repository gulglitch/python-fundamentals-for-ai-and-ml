```python
def add(num1, num2):
    sum = num1+num2
    return sum

def subtract(num1, num2):
  difference = num1 - num2
  return difference

def multiply(num1, num2):
  product=num1*num2
  return product

def divide(num1, num2):
  try:
    quotient=num1/num2
    return quotient
  except ZeroDivisionError:
    print("Error: Cannot divide by zero")
    return None # Or some other indicator of an error

def modulo(num1, num2):
  try:
    remainder=num1%num2
    return remainder
  except ZeroDivisionError:
    print("Error: Cannot perform modulo with zero")
    return None # Or some other indicator of an error

def exponent(num1, num2):
  exponent=num1**num2
  return exponent

def main():
  print("Welcome to the calculator")
  while True: # Main loop for multiple calculations
    while True:
      try:
        num1=float(input("Please enter the first number: "))
        break
      except ValueError:
        print("Invalid input. Please enter a numeric value.")

    while True:
      try:
        num2=float(input("Please enter the second number: "))
        break
      except ValueError:
        print("Invalid input. Please enter a numeric value.")

    while True:
      print("\nPlease select an operation:")
      print("1. Add (+)")
      print("2. Subtract (-)")
      print("3. Multiply (*)")
      print("4. Divide (/)")
      print("5. Modulo (%)")
      print("6. Exponent (**)")
      operation_choice = input("Enter your choice (1-6): ")

      result = None # Initialize result

      if operation_choice == '1':
        result = add(num1, num2)
        break
      elif operation_choice == '2':
        result = subtract(num1, num2)
        break
      elif operation_choice == '3':
        result = multiply(num1, num2)
        break
      elif operation_choice == '4':
        result = divide(num1, num2)
        if result is not None: # Only break if no error occurred
          break
      elif operation_choice == '5':
        result = modulo(num1, num2)
        if result is not None: # Only break if no error occurred
          break
      elif operation_choice == '6':
        result = exponent(num1, num2)
        break
      else:
        print("Invalid choice. Please enter a number between 1 and 6.")

    if result is not None: # Only print result if a calculation was successful
      print(f"Result: {result}")

    while True:
      another_calculation = input("Do you want to perform another calculation? (yes/no): ").lower()
      if another_calculation in ['yes', 'y']:
        break # Continue the main loop
      elif another_calculation in ['no', 'n']:
        print("Goodbye!")
        return # Exit the main loop and the function
      else:
        print("Invalid input. Please enter 'yes' or 'no'.")

main()
```

```python
def filter_positive_numbers(input_file, output_file):
    positive_numbers = []
    try:
        with open(input_file, "r") as file:
            for line in file:
                try:
                    num = float(line.strip())  # Try to convert to float, stripping whitespace
                    if num > 0:                # handles negative numbers
                        positive_numbers.append(num)
                except ValueError:
                    # Skip invalid lines and optionally print a warning
                    print(f"Warning: Skipping non-numeric or empty line: {line.strip()}")           #handles strings
    except FileNotFoundError:
        print(f"Error: Input file '{input_file}' not found.")
        return

    # Now you can write the positive_numbers to the output_file
    try:
        with open(output_file, "w") as file:
            for number in positive_numbers:
                file.write(str(number) + "\n")
    except IOError:
        print(f"Error: Could not write to output file '{output_file}'.")

filter_positive_numbers('/content/input.txt', '/output.txt')
```

```python
class Movie:
  def __init__(self, title, director, release_year, is_rented=False):
    self.title = title
    self.director = director
    self.release_year = release_year
    self.is_rented = is_rented

  def get_details(self):
    if self.is_rented:
        status = "Rented"
    else:
        status = "Available"
    return f"{self.title} ({self.release_year}), directed by {self.director} - {status}"

  def rent(self):
    if self.is_rented:
        print("Sorry, ", self.title, "is currently not available")
    else:
        self.is_rented = True
        print("You have rented ", self.title)

  def return_movie(self):
    if self.is_rented:
        self.is_rented = False
        print("You have returned ", self.title)
    else:
        print("This movie was not rented out")

def main():
  movie1 = Movie("Inception", "Christopher Nolan", 2010)
  print(movie1.get_details()) # "Inception (2010), directed by Christopher Nolan - Available"
  movie1.rent() # "You have rented 'Inception'."
  print(movie1.get_details()) # "Inception (2010), directed by Christopher Nolan - Rented"
  movie1.rent() # "Sorry, 'Inception' is currently not available."
  movie1.return_movie() # "You have returned 'Inception'."
  movie1.return_movie() # "This movie was not rented out."

main()
```
